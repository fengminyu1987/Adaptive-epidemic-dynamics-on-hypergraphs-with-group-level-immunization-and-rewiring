```python
#numpy数组索引 与 Python列表索引

# 关键在于 [2 0 1] 是numpy数组的显示格式（创建时仍需np.array([2, 0, 1])）
# [1, 2, 3] 是Python列表的代码格式（必须用逗号）

import numpy as np

n = 5
states = np.zeros(n, dtype=int)
states1 = np.zeros(n, dtype=int)
states2 = np.zeros(n, dtype=int)

# 方式1：使用numpy数组索引
index_array = np.array([1, 2, 3])
states[index_array] = 1
print("index_array:", index_array)

# 方式2：使用Python列表索引  
index_list = [1, 2, 3]
states1[index_list] = 1
print("index_list:", index_list)

states2[([1, 2, 3])] = 1
states2[[1, 2, 3]] = 1

print("numpy数组索引结果:", states)    # [0 1 1 1 0]
print("Python列表索引结果:", states1)  # [0 1 1 1 0]
print("Python列表索引结果:", states2)  # [0 1 1 1 0]
print("结果相同:", np.array_equal(states, states1))  # True
```

```python
import numpy as np

tri = np.array([2, 5, 1])  # numpy数组
tri.sort()  # 排序后: [1, 2, 5]

result1 = tri.tolist()        # 转换为Python列表
result2 = tuple(tri.tolist()) # 转换为元组

print("tri:", tri, type(tri))
# tri: [1 2 5] <class 'numpy.ndarray'>

# 过程：numpy数组 → 元组
print("tuple(tri):", tuple(tri), type(tuple(tri)))
# tuple(tri): (1, 2, 5) <class 'tuple'>

# 列表 (list) - 不可哈希
print("tri.tolist():", result1, type(result1)) 
# tri.tolist(): [1, 2, 5] <class 'list'>

# 元组 (tuple) - 可哈希
# 过程：numpy数组 → Python列表 → 元组
print("tuple(tri.tolist()):", result2, type(result2))
# tuple(tri.tolist()): (1, 2, 5) <class 'tuple'>
```

```python
import numpy as np

# 假设有5个超边，每个超边的感染节点数
inf_counts = np.array([2, 0, 2, 1, 3])  # 形状: (5,)
print("每个超边的感染节点数:", inf_counts)  # [2 0 2 1 3]

# 找到感染节点数恰好为2的超边
mask = inf_counts == 2
print("布尔掩码:", mask)  # [ True False  True False False]

# 使用np.where
result_tuple = np.where(inf_counts == 2)
print("np.where返回:", result_tuple)  # (array([0, 2]),)

# 提取索引数组
candidates = result_tuple[0]
print("候选超边索引:", candidates)  # [0 2]

print("第一个候选超边", candidates[0]) # 0

states = np.array([1 , 1, 0])

print("states[candidates]", states[candidates])
print("states[candidates[0]]", states[candidates[0]])
```
