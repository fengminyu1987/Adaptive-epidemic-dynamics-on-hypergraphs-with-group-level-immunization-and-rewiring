```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


# 参数设置
out_dir = "wc_experiment_out"  # 与 compute_wc_vs_beta.py 中的 out_dir 一致
out_dir_top = "wc_out_top"
# 检查文件是否存在
summary_file = os.path.join(out_dir, "wc_summary_0.3.csv")
summary_file_top = os.path.join(out_dir_top, "wc_vs_beta_summary_0.3.csv")


summary_df = pd.read_csv(summary_file)
summary_df_top = pd.read_csv(summary_file_top)


# 按 beta 排序
summary_sorted = summary_df_top.sort_values('beta')

# 提取数据
beta_top = summary_sorted['beta'].to_numpy()
mean_wc_top = summary_sorted['mean_wc'].to_numpy()

# 绘制主图
print("\n绘制主图...")
colors = ["#10131F", "#ad2831", '#e09f3e', "#7FBD53",]
markers = ['o', 's', '^', 'D']
markerfacecolors = ['#03071e', "#e42e2e", "#e77903", "#38b000",]
markeredgecolors = ['#03071e', "#e42e2e", '#e77903', "#38b000",]

plt.figure(figsize=(9,6))


# 绘制 Top 线
plt.plot(beta_top, mean_wc_top, linestyle='--', marker='*', 
                    markerfacecolor='#0466c8', markeredgecolor='#0466c8',
                    linewidth=2.5, markersize=12, label=r'$\mathrm{TI}$', color='#2589bd')

# 构建标签顺序：theta_del 数值升序 + random 
unique_keys = []

# random
unique_keys.append(('random', np.nan))

# numeric theta
numeric_keys = sorted([k for k in summary_df['theta_del'].unique() if not k == -1])
for th in numeric_keys:
    unique_keys.append(('threshold', float(th)))

for i, key in enumerate(unique_keys):
    del_type, th = key
    if del_type == 'random':
        sub = summary_df[summary_df['deletion_type'] == 'random']
        label = r'$\mathrm{Random}$'
    else:
        sub = summary_df[(summary_df['deletion_type']=='threshold') & (np.isclose(summary_df['theta_del'], th))]
        # 根据 theta_del 值分配 SI_1, SI_2, SI_3 标签
        if th <= 0.2:
            label = r'$\mathrm{SI_1}$'
        elif th <= 0.4:
            label = r'$\mathrm{SI_2}$'
        else:
            label = r'$\mathrm{SI_3}$'
    if sub.empty:
        continue
    sub = sub.sort_values('beta')
    x = sub['beta'].to_numpy()
    y = sub['mean_wc'].to_numpy()
    color = colors[i % len(colors)]
    marker = markers[i % len(markers)]
    markerfacecolor = markerfacecolors[i % len(markerfacecolors)]
    markeredgecolor = markeredgecolors[i % len(markeredgecolors)]

    plt.plot(x, y, marker=marker, color=color, linewidth=2.5, markerfacecolor = markerfacecolor, markeredgecolor = markeredgecolor,label=label, markersize=12, linestyle='--',)

plt.xlabel(r' $\beta$', fontsize=30)
plt.ylabel(r'$w_c$', fontsize=30)
# plt.title(r'HIT $w_c$ vs $\beta$ for different deletion strategies', fontsize=14)
plt.ylim(0.0, 1.0)
plt.xlim(0.245, 0.4)
plt.xticks(np.linspace(0.25,0.40,6), fontsize=25)
plt.yticks(np.linspace(0.0, 1.0,6), fontsize=25)
# plt.xlim(min(summary_df['beta']), max(summary_df['beta']))
plt.grid(alpha=0.2, linestyle='--', linewidth=0.8)

legend = plt.legend(
    loc='lower right',
    # bbox_to_anchor=(1, 0.55),
    fontsize='18',
    handletextpad=0.3,    # 图标与文本的间距
    #columnspacing=1,      # 列间距（多列时有效）
    borderpad=0.4,          # 图例内边距
    # ncol=2,              # 分两列
    handlelength=2.0,    # 统一图标宽度
    columnspacing=0.5,     # 缩小列间距（默认2.0）
)
legend.get_frame().set_edgecolor('grey')
legend.get_frame().set_linewidth(1)

# 启用四面刻度
plt.tick_params(
    axis='both',      # 同时控制 x 和 y 轴
    which='both',     # 同时控制主刻度和次刻度
    top=True,         # 显示顶部刻度
    right=True,       # 显示右侧刻度
    labeltop=False,   # 不显示顶部刻度标签（避免重叠）
    labelright=False,  # 不显示右侧刻度标签（避免重叠）
    direction='in',   # 刻度朝内
)


plt.tight_layout()
fig_fn = os.path.join("./", "wc_vs_beta_0.3.pdf")
plt.savefig(fig_fn, dpi=300, bbox_inches='tight')
print(f"[INFO] 已保存图像 -> {fig_fn}")
plt.show()

```
